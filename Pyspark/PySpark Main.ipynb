```python
from pyspark.sql.types import *


data = [
    ("sai", 7000, "IT"),
    ("pradeep", 8000, "FC"),
    ("rahul", 7500, "IT"),
    ("anita", 7200, "HR"),
    ("kiran", 6800, "IT"),
    ("meena", 8100, "Finance"),
    ("ravi", 7700, "FC"),
    ("divya", 7300, "HR"),
    ("arjun", 7900, "IT"),
    ("pooja", 7600, "Finance")
]

schema = StructType([
    StructField("name", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("department", StringType(), True)
])
```

```python
df=spark.createDataFrame(data,schema)
display(df)
```

```python
from pyspark.sql.functions import spark_partition_id

df_with_partitions = df.withColumn("partition_id", spark_partition_id())

df.repartition(5)
display(df)
```

```python
df.filter(df.salary>7500).show()
df.where("Salary=7600").show()
```

```python
from pyspark.sql.functions import *
bonus=10
bonus=bonus/100
df=df.withColumn("Bonus",col("salary")*bonus)
display(df)
```

```python
df.distinct().show
df.dropDuplicates().show()
```

```python
df.orderBy(col("salary").desc()).show()
```

```python
df.groupBy("Department").agg(sum("salary").alias("Total")).orderBy(col("Total").desc()).show()
```

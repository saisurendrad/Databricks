```python
import requests
from pyspark.sql.functions import *
from pyspark.sql import types
from pyspark.sql.functions import col

```

```python
response = requests.get('https://dummyjson.com/products')
if response.status_code == 200:
    df = response.json()
    display(df)
else:
    print(f"API request failed with status {response.status_code}\n{response.text[:500]}")
```

```python
schema = types.StructType([
    types.StructField('id', types.IntegerType()),
    types.StructField('title', types.StringType()),
    types.StructField('description', types.StringType()),
    types.StructField('price', types.DoubleType())
])
```

```python
df = spark.createDataFrame(df['products'], schema)
display(df)
```

```python
df_aggregated = df.groupBy("price").agg(
    sum(col("price")).alias("total_price")
).dropDuplicates().orderBy(col("total_price").desc())

display(df_aggregated)
```

```python
from pyspark.sql.window import Window

df = df.withColumn(
    "rank",
    row_number().over(Window.partitionBy("id").orderBy(col("price").desc()))
)
display(df)
```
